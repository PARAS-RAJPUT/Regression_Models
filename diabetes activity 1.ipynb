```python
# Separate the independent  (X) and dependent (y) variables

X = df.iloc[:, :-1].values #all colimns except last column
y = df.iloc[:, -1].values #outcome columns
X[:5] #


```

```python
#feature scaling(optional but improves model convergence)

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
X_train
```

```python
#Train the logistic regression model

classifier = LogisticRegression()
classifier.fit(X_train, y_train)
```

```python
#Predict test results

y_pred = classifier.predict(X_test)
print(y_pred, y_test)
```

```python
#Evaluate model performance

cm = confusion_matrix(y_test, y_pred)
print(cm)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
#Predict for a new sample

new_data = np.array([[150]])  # Only Glucose value
new_sc = sc.fit_transform(new_data)
prediction1 = classifier.predict(new_sc)
print("Prediction:", prediction1)

```
