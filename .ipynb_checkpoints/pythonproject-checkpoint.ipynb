```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_excel("AviationAccidentStatistics_2004-2023_20241217.xlsx")
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')  # Normalize column names
df.dropna(how='all', inplace=True)  # Drop rows with all NaNs
df.info()
```

```python
df['year'] = pd.DatetimeIndex(df['date']).year
yearly_accidents = df.groupby('year').size()

plt.figure(figsize=(10,5))
sns.lineplot(x=yearly_accidents.index, y=yearly_accidents.values, marker='o')
plt.title('Yearly Aviation Accidents')
plt.xlabel('Year')
plt.ylabel('Number of Accidents')
plt.grid(True)
plt.show()
```

```python
df['is_fatal'] = np.where(df['fatalities'] > 0, 'Fatal', 'Non-Fatal')
fatal_counts = df['is_fatal'].value_counts()

plt.figure(figsize=(6,6))
fatal_counts.plot.pie(autopct='%1.1f%%', startangle=90, colors=['tomato', 'skyblue'])
plt.title('Fatal vs Non-Fatal Accidents')
plt.ylabel('')
plt.show()
```

```python
top_states = df['location_state'].value_counts().nlargest(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_states.values, y=top_states.index, palette='viridis')
plt.title('Top 10 States with Most Aviation Accidents')
plt.xlabel('Number of Accidents')
plt.ylabel('State')
plt.show()
```
